#1. DATASET OVERVIEW

```python
# 1. Dataset Overview
import pandas as pd
import numpy as np
from pathlib import Path

# Dataset directory
DATASET_PATH = Path("../content")

# Load datasets
datasets = {
    "customers": pd.read_csv(DATASET_PATH / "olist_customers_dataset.csv"),
    "geolocation": pd.read_csv(DATASET_PATH / "olist_geolocation_dataset.csv"),
    "order_items": pd.read_csv(DATASET_PATH / "olist_order_items_dataset.csv"),
    "order_payments": pd.read_csv(DATASET_PATH / "olist_order_payments_dataset.csv"),
    "order_reviews": pd.read_csv(DATASET_PATH / "olist_order_reviews_dataset.csv"),
    "orders": pd.read_csv(DATASET_PATH / "olist_orders_dataset.csv"),
    "products": pd.read_csv(DATASET_PATH / "olist_products_dataset.csv"),
    "sellers": pd.read_csv(DATASET_PATH / "olist_sellers_dataset.csv"),
    "category_translation": pd.read_csv(DATASET_PATH / "product_category_name_translation.csv")
}
```

```python
for table_name, df in datasets.items():

    print("="*80)
    print(table_name.upper())
    print("="*80)

    display(
        pd.DataFrame({
            "Column": df.columns,
            "Data Type": df.dtypes.astype(str)
        })
    )
```

```python
for table_name, df in datasets.items():

    print("="*80)
    print(table_name.upper())
    print("="*80)

    display(df.head())
```

```python
summary = []

for table_name, df in datasets.items():

    summary.append({
        "Table": table_name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Memory (MB)": round(df.memory_usage(deep=True).sum()/1024**2,2)
    })

dataset_summary = pd.DataFrame(summary)

dataset_summary
```

```python
for table_name, df in datasets.items():

    object_columns = df.select_dtypes(include="object").columns

    if len(object_columns) == 0:
        continue

    print("=" * 100)
    print(f"{table_name.upper()} - CATEGORICAL VALUE DISTRIBUTION")
    print("=" * 100)

    for column in object_columns:

        print(f"\nColumn: {column}")

        display(
            df[column]
            .value_counts(dropna=False)
            .rename_axis("Value")
            .reset_index(name="Count")
        )
```

**Findings**

- The dataset contains all major business entities of an e-commerce platform, including Customer, Order, Product, Seller, Payment, and Review.
- Nearly 100,000 customers made purchases between 2016 and 2018 across more than 8,000 cities in Brazil.
- The `product_category_name_translation` table provides English translations for product category names, enabling easier business analysis and reporting.
- The `geolocation` table is the largest dataset, containing over one million geographic location records.
- The dataset includes both transactional data and reference data, providing a solid foundation for building Dimension Tables in a Data Warehouse.

#2. Data Quality

```python
# 2. Data Quality Assessment
missing_summary = []

for table_name, df in datasets.items():

    total_rows = len(df)

    for column in df.columns:

        missing = df[column].isna().sum()

        missing_summary.append({

            "Table": table_name,

            "Column": column,

            "Missing Count": missing,

            "Missing Rate (%)": round(missing/total_rows*100,2)

        })

missing_summary = pd.DataFrame(missing_summary)

missing_summary.sort_values(
    "Missing Rate (%)",
    ascending=False
)
```

```python
duplicate_summary = []

for table_name, df in datasets.items():

    duplicate_summary.append({

        "Table": table_name,

        "Total Rows": len(df),

        "Duplicate Rows": df.duplicated().sum(),

        "Duplicate Rate (%)":
        round(df.duplicated().mean()*100,2)

    })

duplicate_summary = pd.DataFrame(duplicate_summary)

duplicate_summary
```

```python
quality_summary = []

for table_name, df in datasets.items():

    quality_summary.append({

        "Table": table_name,

        "Rows": len(df),

        "Columns": len(df.columns),

        "Missing Values": df.isna().sum().sum(),

        "Duplicate Rows": df.duplicated().sum()

    })

quality_summary = pd.DataFrame(quality_summary)

quality_summary
```

**Findings**

- Missing values are primarily concentrated in the `order_reviews` table.
- Duplicate records are found only in the `geolocation` table.
- No duplicate records were detected in the main transactional tables, including `orders`, `order_items`, and `order_payments`.
- Primary identifiers (`customer_id`, `order_id`, `product_id`, and `seller_id`) contain no missing values, ensuring reliable relationships between the core business entities.

```python
# 3. Relational Data Assessment
relationships = [
    ("customers", "customer_id", "orders", "customer_id"),
    ("orders", "order_id", "order_items", "order_id"),
    ("orders", "order_id", "order_payments", "order_id"),
    ("orders", "order_id", "order_reviews", "order_id"),
    ("products", "product_id", "order_items", "product_id"),
    ("sellers", "seller_id", "order_items", "seller_id"),
    (
        "category_translation",
        "product_category_name",
        "products",
        "product_category_name",
    ),
]

relationship_df = pd.DataFrame(
    relationships,
    columns=[
        "Parent Table",
        "Primary Key",
        "Child Table",
        "Foreign Key",
    ],
)

relationship_df
```

```python
primary_keys = {
    "customers": "customer_id",
    "orders": "order_id",
    "products": "product_id",
    "sellers": "seller_id",
    "order_reviews": "review_id",
    "category_translation": "product_category_name",
}

pk_summary = []

for table, pk in primary_keys.items():

    df = datasets[table]

    pk_summary.append({

        "Table": table,

        "Primary Key": pk,

        "Null": df[pk].isna().sum(),

        "Duplicate": df[pk].duplicated().sum()

    })

pd.DataFrame(pk_summary)
```

```python
cardinality_summary = []

for parent_table, pk, child_table, fk in relationships:

    child = datasets[child_table]

    cardinality_summary.append({

        "Relationship":
        f"{parent_table} → {child_table}",

        "Unique Parent":
        child[fk].nunique(),

        "Total Child Rows":
        len(child)

    })

pd.DataFrame(cardinality_summary)
```

**Findings**

- Most tables exhibit one-to-many (1:N) relationships, while the `customers` and `orders` tables have a one-to-one (1:1) relationship based on `customer_id`.
- No orphan records were detected among the core business tables, indicating that the dataset maintains good referential integrity.
